TASK 3: Analyze customer orders

```python
# Customer Orders
#use a loop to calculate the total amount each customer spends. >100-high Value buyer, <100-low value buyer, >50 to <100- Moderate buyer
orders = [
("Alice", "Laptop", 800, "Electronics"),
("Bob", "T-Shirt", 25, "Clothing"),
("Charlie", "Mixer", 60, "Home Essentials"),
("David", "Headphones", 45, "Electronics"),
("Emma", "Jacket", 75, "Clothing")
]
# Calculate spending and classify customers
for name, product, price, category in orders:
    if price > 100:
        status = "High-Value Buyer"
    elif price >= 50:
        status = "Moderate Buyer"
    else:
        status = "Low-Value Buyer"
    print(name, "- $", price, "-", status)
```

TASK 4: Generate business insights

```python
orders = [
("Alice", "Laptop", 800, "Electronics"),
("Bob", "T-Shirt", 25, "Clothing"),
("Charlie", "Mixer", 60, "Home Essentials"),
("David", "Headphones", 45, "Electronics"),
("Emma", "Jacket", 75, "Clothing")
]
# 1. Revenue per category
revenue = {}
for name, product, price, category in orders:
    revenue[category] = revenue.get(category, 0) + price
print("Revenue:", revenue)
```

```python
# 2. Unique products
products = {product for name, product, price, category in orders}
print("Unique Products:", products)
```

```python
# 3. Customers who purchased Electronics
electronics_customers = [name for name, product, price, category in orders
if category == "Electronics"]
print("Electronics Customers:", electronics_customers)
```

```python
# 4. Top 3 highest-spending customers
top_customers = sorted([(name, price) for name, product, price, category in orders],
key=lambda x: x[1], reverse=True)[:3]
print("Top 3 Customers:", top_customers)
```

TASK 5: Organize and display data

```python
#Print a summary of each customer’s total spending and their classification
for name, product, price, category in orders:
    if price > 100:
        status = "High-Value Buyer"
    elif price >= 50:
        status = "Moderate Buyer"
    else:
        status = "Low-Value Buyer"
    print(name, "-", price, "-", status)
```

```python
# Customers by category
electronics = {name for name, p, price, cat in orders if cat == "Electronics"}
clothing = {name for name, p, price, cat in orders if cat == "Clothing"}
home = {name for name, p, price, cat in orders if cat == "Home Essentials"}

# Customers with multiple categories
multiple_categories = (electronics & clothing) | (electronics & home) | (clothing & home)
print("\nMultiple Category Customers:", multiple_categories)

# Customers who bought both Electronics and Clothing
common_customers = electronics & clothing
print("Electronics and Clothing Customers:", common_customers)
```

```python
#FINAL REPORT
print("Total categories:", categories)
print("revenue by categories:", revenue)
print("Top customers:", top_customers)
print("multi category customers:", multiple_categories)
print("common electronic and clothing buyer:", common_customers)
```
